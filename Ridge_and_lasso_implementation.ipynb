```python
import pandas as pd
```

```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

```python
data_url = "http://lib.stat.cmu.edu/datasets/boston"
raw_df = pd.read_csv(data_url, sep="\s+", skiprows=22, header=None)
data = np.hstack([raw_df.values[::2, :], raw_df.values[1::2, :2]])
target = raw_df.values[1::2, 2]
```

```python
datasets=pd.DataFrame(data)
```

```python
feature_names = ['CRIM', 'ZN', 'INDUS', 'CHAS', 'NOX', 'RM', 'AGE', 'DIS', 'RAD', 'TAX', 'PTRATIO', 'B', 'LSTAT']
datasets.columns = feature_names
```

```python
datasets.head()
```

```python
## independent and dependents features
X=datasets
Y=target
```

```python
Y
```

```python
## Train and Test Split
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.3,random_state=42)
```

```python
X_test
```

```python
from sklearn.linear_model import LinearRegression
regression = LinearRegression()
```

```python
#standardize the dataset
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
X_train=scaler.fit_transform(X_train)
```

```python
X_test =scaler.transform(X_test)
#To prevent data leakage we are using transform in test data so that the same mean and SD is used to standardize the dataset
```

```python
#cross validation
from sklearn.model_selection import cross_val_score
```

```python
regression=LinearRegression()
regression.fit(X_train,Y_train)
```

```python
mse=cross_val_score(regression,X_train,Y_train,scoring='neg_mean_squared_error',cv=5)
```

```python
np.mean(mse)
```

```python
#prediction
reg_pred=regression.predict(X_test)
```

```python
import seaborn as sns
sns.displot(reg_pred-Y_test,kind='kde')
```

```python
from sklearn.metrics import r2_score
```

```python
score=r2_score(Y_test, reg_pred)
```

```python
score
```

Ridge Regression

```python
from sklearn.linear_model import Ridge
from sklearn.model_selection import GridSearchCV
```

```python
ridge_regressor=Ridge()
```

```python
parameters={'alpha':[1,2,5,6,7,9,23,56,78,88]}
ridgecv=GridSearchCV(ridge_regressor,parameters,scoring='neg_mean_squared_error',cv=5)
ridgecv.fit(X_train,Y_train)
```

```python
print(ridgecv.best_params_)
```

```python
print(ridgecv.best_score_)
```

```python
ridge_pred=ridgecv.predict(X_test)
```

```python
import seaborn as sns
sns.displot(ridge_pred-Y_test,kind='kde')
```

```python
score=r2_score(Y_test, ridge_pred)
```

```python
score
```

Lasso regression

```python
from sklearn.linear_model import Lasso
```

```python
lasso=Lasso()
```

```python
parameters={'alpha':[1,2,5,6,7,9,23,56,78,88]}
lassocv=GridSearchCV(lasso,parameters,scoring='neg_mean_squared_error',cv=5)
lassocv.fit(X_train,Y_train)
```

```python
print(lassocv.best_params_)
print(lassocv.best_score_)
```

```python
lasso_pred=lassocv.predict(X_test)
```

```python
import seaborn as sns
sns.displot(lasso_pred-Y_test,kind='kde')
```

```python
score=r2_score(Y_test, lasso_pred)
```
